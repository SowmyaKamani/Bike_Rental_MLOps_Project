```python
%%writefile ../src/api.py

from fastapi import FastAPI
from pydantic import BaseModel
import pandas as pd
import mlflow

# --------------------------------------------------
# FastAPI application
# --------------------------------------------------

app = FastAPI(
    title="Bike Rental Demand Prediction API",
    description="API for predicting bike rental demand",
    version="1.0.0"
)

# --------------------------------------------------
# MLflow configuration
# --------------------------------------------------

# IMPORTANT:
# Uvicorn is started from the project root,
# so mlflow.db is in the current directory.
mlflow.set_tracking_uri("sqlite:///./mlflow.db")

MODEL_NAME = "Bike_Rental_Demand_Model"
MODEL_ALIAS = "champion"

# Load the champion model from MLflow Model Registry
model = mlflow.pyfunc.load_model(
    f"models:/{MODEL_NAME}@{MODEL_ALIAS}"
)

print("MLflow model loaded successfully.")
print("Model:", MODEL_NAME)
print("Alias:", MODEL_ALIAS)


# --------------------------------------------------
# Input schema
# --------------------------------------------------

class BikeRentalInput(BaseModel):

    DayOfWeek: str
    Season: str
    Weather: str

    Temperature: float
    Humidity: float
    WindSpeed: float
    RentalDuration: float

    UserType: str
    PaymentMethod: str
    Cancelled: int

    Year: int
    Month: int
    Day: int
    DayOfWeek_Num: int
    Hour: int
    Is_Weekend: int

    Month_Sin: float
    Month_Cos: float
    Hour_Sin: float
    Hour_Cos: float


# --------------------------------------------------
# Root endpoint
# --------------------------------------------------

@app.get("/")
def home():

    return {
        "message": "Bike Rental Demand Prediction API",
        "model": MODEL_NAME,
        "alias": MODEL_ALIAS
    }


# --------------------------------------------------
# Prediction endpoint
# --------------------------------------------------

@app.post("/predict")
def predict(data: BikeRentalInput):

    # Convert request to dictionary
    input_data = data.model_dump()

    # Convert to DataFrame
    input_df = pd.DataFrame([input_data])

    # Make prediction
    prediction = model.predict(input_df)

    return {
        "prediction": float(prediction[0])
    }
```
